```python
import re

MASTER_REGEX = re.compile(
    r'(?P<url>https?://[^\s]+|www\.[^\s]+)'                    # URLs
    r'|(?P<email>[\w.+-]+@[\w-]+\.[\w.-]+)'                     # Emails
    r'|(?P<date>\d{1,2}[-/]\d{1,2}[-/]\d{2,4})'                 # Dates: DD-MM-YYYY
    r'|(?P<comma_num>\d{1,2}(?:,\d{2})*,\d{3})'                 # Indian & Western comma numbers
    r'|(?P<decimal>\d+\.\d+)'                                   # Decimals: 3.14
    r'|(?P<num_suffix>\d+[\u0C00-\u0C7F]+)'                     # Number + Telugu suffix: 18న
    r'|(?P<number>\d+)'                                         # Plain integers
    r'|(?P<word>[a-zA-Z\u0C00-\u0C7F\u200C\u200D]+)'            # Telugu/English words (FIXED)
    r'|(?P<punct>[^\s])'                                        # Anything else = punctuation/symbol
)

def word_tokenize(text):
    """Returns list of (token, label) tuples for one sentence/text chunk."""
    return [(m.group(0), m.lastgroup) for m in MASTER_REGEX.finditer(text)]

# Quick test
test = "ఇది కూడా చదవండి: Gold rate 8-5-2021: 1,00,000 పెరిగిన బంగారం ధర 9.80 లక్షలు"
for tok, label in word_tokenize(test):
    print(f"{label:12s} -> {tok}")
```

```python
PROTECTED_GROUPS = {"url", "email", "date", "decimal", "comma_num", "num_suffix"}

TELUGU_ABBREVIATIONS = [
    "క్రీ.శ.", "క్రీ.పూ.",
    "కి.మీ.", "కి. మీ.",
    "మి.మీ.", "మి. మీ.",
    "డా.", "ప్రొ.", "శ్రీ.", "నం.", "సం.",
    "Dr.", "Mr.", "Mrs.", "Prof.", "Ms.", "U.S.", "U.K.",
]
# Escape for regex safety, longest-first (already sorted above), join as alternation
ABBREV_REGEX = re.compile(
    "|".join(re.escape(a) for a in TELUGU_ABBREVIATIONS)
)

SENTENCE_END_CHARS = r'[.?!\u2026]'

CLOSING_QUOTES = '"\'\u201d\u2019'  # " ' " '

def sentence_tokenize(text):
    """Split text into sentences, respecting abbreviations, protected numeric/URL/date
    patterns, and closing quotation marks."""

    # 1. Protected spans from MASTER_REGEX (url/email/date/decimal/comma_num/num_suffix)
    protected_spans = [
        (m.start(), m.end())
        for m in MASTER_REGEX.finditer(text)
        if m.lastgroup in PROTECTED_GROUPS
    ]
    # 2. Add abbreviation spans as protected too
    protected_spans += [(m.start(), m.end()) for m in ABBREV_REGEX.finditer(text)]

    def inside_protected(pos):
        return any(start < pos <= end for start, end in protected_spans)

    # 3. Candidate boundary positions: sentence-ending punctuation not inside a protected span
    boundary_positions = []
    for m in re.finditer(SENTENCE_END_CHARS, text):
        pos = m.end()
        if inside_protected(pos):
            continue
        # 4. Swallow any immediately following closing quote(s) into this sentence
        while pos < len(text) and text[pos] in CLOSING_QUOTES:
            pos += 1
        boundary_positions.append(pos)

    # 5. Slice text at the real boundaries
    sentences = []
    start = 0
    for pos in boundary_positions:
        sentence = text[start:pos].strip()
        if sentence:
            sentences.append(sentence)
        start = pos
    remainder = text[start:].strip()
    if remainder:
        sentences.append(remainder)

    return sentences

# --- Test against real problem cases from your data ---
test_cases = [
    "సబ్ పోస్టాఫీసు సౌకర్యం గ్రామానికి 5 కి.మీ. లోపు దూరంలో ఉంది. పోస్టాఫీసు సౌకర్యం 10 కి.మీ.కి పైబడిన దూరంలో ఉంది.",
    "ఇది మండల కేంద్రమైన హథ్నూర నుండి 16 కి. మీ. దూరం లోను ఉంది. 2011 జనగణన ప్రకారం.",
    "ఈ గ్రామంలో \"శ్రీ నెట్టికంటి ఆంజనేయ స్వామి\" వారు కొలువై వున్నారు. భక్తులు వస్తుంటారు.",
    "నమస్తే తెలంగాణ ప్రత్యేక కథనం… ఇది కొనసాగుతుంది.",
]

for t in test_cases:
    print("INPUT:", t)
    for s in sentence_tokenize(t):
        print("  ->", s)
    print()


para = "సెప్టెంబర్‌లో 9.80 లక్షల వరద. ఏప్రిల్ 2న శ్రీరామ నవమి జరుగుతుంది! మీకు తెలుసా?"
for s in sentence_tokenize(para):
    print("-", s)
```

```python
def tokenize_paragraph(paragraph):
    """
    Returns a list of tokenized sentences (each a list of word tokens),
    and also a list of space-joined tokenized sentence strings (for file writing).
    """
    sentences = sentence_tokenize(paragraph)
    tokenized_sentences = []
    for sent in sentences:
        tokens = [tok for tok, label in word_tokenize(sent)]
        tokenized_sentences.append(tokens)
    return tokenized_sentences

def paragraph_to_output_lines(paragraph):
    """Returns list of strings, one per sentence, tokens space-joined (assignment's required format)."""
    tokenized_sentences = tokenize_paragraph(paragraph)
    return [" ".join(tokens) for tokens in tokenized_sentences]

# Test on a real line from your cleaned data
sample_para = cleaned_lines[0]
print("ORIGINAL:", sample_para)
print()
print("TOKENIZED OUTPUT LINES:")
for line in paragraph_to_output_lines(sample_para):
    print(" ", line)
```

```python
import os
import re
import json
import unicodedata
from tqdm import tqdm
import pyarrow as pa
import pyarrow.parquet as pq

# ==========================================
# 1. CONSTANTS & REGEX COMPILES
# ==========================================
INPUT_FILE = "/content/drive/MyDrive/NLP/Assignment_1/indiccorp_telugu.txt"
OUTPUT_DIR = "/content/drive/MyDrive/NLP/Assignment_1/processed"
os.makedirs(OUTPUT_DIR, exist_ok=True)

PARQUET_FILE = os.path.join(OUTPUT_DIR, "telugu_tokenized_full.parquet")
TXT_FILE = os.path.join(OUTPUT_DIR, "telugu_tokenized_full.txt")
STATS_FILE = os.path.join(OUTPUT_DIR, "telugu_corpus_stats.json")

TELUGU_DIGITS = "౦౧౨౩౪౫౬౭౮౯"
ARABIC_DIGITS = "0123456789"
DIGIT_MAP = str.maketrans(TELUGU_DIGITS, ARABIC_DIGITS)

MASTER_REGEX = re.compile(
    r'(?P<url>https?://[^\s]+|www\.[^\s]+)'
    r'|(?P<email>[\w.+-]+@[\w-]+\.[\w.-]+)'
    r'|(?P<date>\d{1,2}[-/]\d{1,2}[-/]\d{2,4})'
    r'|(?P<comma_num>\d{1,2}(?:,\d{2})*,\d{3})'
    r'|(?P<decimal>\d+\.\d+)'
    r'|(?P<num_suffix>\d+[\u0C00-\u0C7F]+)'
    r'|(?P<number>\d+)'
    r'|(?P<word>[a-zA-Z\u0C00-\u0C7F\u200C\u200D]+)'
    r'|(?P<punct>[^\s])'
)

PROTECTED_GROUPS = {"url", "email", "date", "decimal", "comma_num", "num_suffix"}

TELUGU_ABBREVIATIONS = [
    "క్రీ.శ.", "క్రీ.పూ.", "కి.మీ.", "కి. మీ.", "మి.మీ.", "మి. మీ.",
    "డా.", "ప్రొ.", "శ్రీ.", "నం.", "సం.", "Dr.", "Mr.", "Mrs.", "Prof.", "Ms.", "U.S.", "U.K."
]
ABBREV_REGEX = re.compile("|".join(re.escape(a) for a in TELUGU_ABBREVIATIONS))
SENTENCE_END_CHARS = r'[.!?\u2026]'
CLOSING_QUOTES = '"\'\u201d\u2019'

# ==========================================
# 2. PROCESSING FUNCTIONS
# ==========================================
def clean_text(text):
    """Combines all previous map functions into one."""
    text = text.strip()
    if not text:
        return ""
    text = unicodedata.normalize('NFC', text)
    text = re.sub(r'\s+', ' ', text).strip()
    text = text.translate(DIGIT_MAP)
    text = re.sub(r'([.!?])\1+', r'\1', text)
    return text

def word_tokenize(text):
    return [(m.group(0), m.lastgroup) for m in MASTER_REGEX.finditer(text)]

def sentence_tokenize(text):
    protected_spans = [(m.start(), m.end()) for m in MASTER_REGEX.finditer(text) if m.lastgroup in PROTECTED_GROUPS]
    protected_spans += [(m.start(), m.end()) for m in ABBREV_REGEX.finditer(text)]

    def inside_protected(pos):
        return any(start < pos <= end for start, end in protected_spans)

    boundary_positions = []
    for m in re.finditer(SENTENCE_END_CHARS, text):
        pos = m.end()
        if inside_protected(pos):
            continue
        while pos < len(text) and text[pos] in CLOSING_QUOTES:
            pos += 1
        boundary_positions.append(pos)

    sentences = []
    start = 0
    for pos in boundary_positions:
        sentence = text[start:pos].strip()
        if sentence:
            sentences.append(sentence)
        start = pos
    remainder = text[start:].strip()
    if remainder:
        sentences.append(remainder)

    return sentences

# ==========================================
# 3. STREAMING PIPELINE (RAM-SAFE)
# ==========================================
# Initialize PyArrow Parquet Writer for appending batches
schema = pa.schema([('tokenized_sentence', pa.string())])
parquet_writer = pq.ParquetWriter(PARQUET_FILE, schema, compression='snappy')

# Statistics Tracking
total_sentences = 0
total_words = 0
total_chars = 0
unique_tokens = set()

BATCH_SIZE = 100_000  # Save to disk every 100k sentences
batch_sentences = []

print("Starting full corpus processing. This will take a while...")

# We know the total lines is ~21,458,261 from your previous output, which makes tqdm accurate
TOTAL_LINES = 21_458_261

with open(INPUT_FILE, "r", encoding="utf-8") as f_in, \
     open(TXT_FILE, "w", encoding="utf-8") as f_txt_out:

    for line in tqdm(f_in, total=TOTAL_LINES, desc="Processing Lines"):
        cleaned = clean_text(line)
        if not cleaned:
            continue

        sentences = sentence_tokenize(cleaned)

        for sent in sentences:
            word_tokens = word_tokenize(sent)
            if not word_tokens:
                continue

            total_sentences += 1
            sentence_words = []

            for tok, label in word_tokens:
                sentence_words.append(tok)
                unique_tokens.add(tok)

                # Check if word contains alphanumeric characters (to exclude pure punctuation from word/char counts)
                if any(ch.isalnum() for ch in tok):
                    total_words += 1
                    total_chars += len(tok)

            # Form space-separated sentence string
            joined_sentence = " ".join(sentence_words)
            batch_sentences.append(joined_sentence)
            f_txt_out.write(joined_sentence + "\n")

            # Write Parquet batch when full to save RAM
            if len(batch_sentences) >= BATCH_SIZE:
                table = pa.Table.from_arrays([pa.array(batch_sentences)], schema=schema)
                parquet_writer.write_table(table)
                batch_sentences = []

    # Write remaining sentences in the final batch
    if batch_sentences:
        table = pa.Table.from_arrays([pa.array(batch_sentences)], schema=schema)
        parquet_writer.write_table(table)

parquet_writer.close()

# ==========================================
# 4. FINAL STATISTICS
# ==========================================
avg_sentence_length = total_words / total_sentences if total_sentences > 0 else 0
avg_word_length = total_chars / total_words if total_words > 0 else 0
ttr = len(unique_tokens) / (total_words)

stats = {
    "Total number of sentences": total_sentences,
    "Total number of words": total_words,
    "Total number of characters": total_chars,
    "Average Sentence Length (words/sentence)": round(avg_sentence_length, 3),
    "Average word length (characters/word)": round(avg_word_length, 3),
    "Type/Token Ratio (TTR)": round(ttr, 6),
    "Total unique tokens": len(unique_tokens)
}

print("\n--- CORPUS STATISTICS ---")
for k, v in stats.items():
    print(f"{k}: {v}")

with open(STATS_FILE, "w", encoding="utf-8") as f_stats:
    json.dump(stats, f_stats, indent=4, ensure_ascii=False)

print(f"\nProcessing complete! Files saved to: {OUTPUT_DIR}")
```
